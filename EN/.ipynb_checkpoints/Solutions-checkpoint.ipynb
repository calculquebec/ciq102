```python
import pennylane as qp
import numpy as np
```

```python
#Exercice 1
dev = qp.device("default.qubit", wires = 2)
@qp.qnode(dev)
def circuit():
    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0,1])
    return qp.state()
```

```python
#Exercice 2
#a)
dev = qp.device('default.qubit', wires = 2)
@qp.qnode(dev)
def phi_moins():
    qp.PauliX(wires = 0)
    qp.Hadamard(wires = 0)
    qp.CNOT(wires = [0,1])
    return qp.state()

#b)
@qp.qnode(dev)
def psi_plus():
  qp.Hadamard(wires = 0)
  qp.CNOT(wires = [0,1])
  qp.PauliX(wires = 0)
  return qp.state()

#c)
@qp.qnode(dev)
def psi_moins():
  qp.Hadamard(wires = 0)
  qp.CNOT(wires = [0,1])
  qp.Z(wires = 0)
  qp.X(wires = 1)
  return qp.state()
```

```python
#Exercice 3
dev = qp.device('default.qubit', wires=1)
@qp.qnode(dev)
def prepare_with_gate():
  qp.RX(np.pi/3, wires = 0)
  return qp.state()

state = np.array([np.sqrt(3)/2, -0.5j])

@qp.qnode(dev)
def prepare_with_StatePrep(state):
  qp.StatePrep(state=state, wires=0)
  return qp.state()
```

```python
#Exercice 4
dev = qp.device('default.qubit', wires = 1)
@qp.qnode(dev)
def circuit():

    qp.Hadamard(wires = 0)
    qp.Z(wires = 0)

    return qp.expval(qp.Z(wires = 0))
    
print(circuit())
```

```python
#Exercice 6
dev = qp.device("default.qubit", wires=1)

@qp.qnode(dev)
def variational_circuit(observable, params):
    """
    Implement a simple variational circuit with two rotations:
    Rx and Ry, whose angles are given in params.

    Args:
        observable (qp.operation.Observable): The observable to measure.
        params (list[float]): List of parameters (theta_x, theta_y).

    Returns:
        float: The average value <observable>.
    """
    qp.RX(params[0], wires=0) # Your code here 
    qp.RY(params[1], wires=0) # Your code here
    
    return qp.expval(observable)

#  Example test with the PauliZ observable
params = np.array([0.1, 0.2], requires_grad=True)
print("Result =", variational_circuit(qp.PauliZ(0), params))
```
